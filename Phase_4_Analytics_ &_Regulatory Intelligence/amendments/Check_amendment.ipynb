```python
import pandas as pd
from pathlib import Path

file_path = Path("validated_output/T1001_v2_validated.csv")

df = pd.read_csv(file_path)

print("Total records:", len(df))
print("Total columns:", len(df.columns))
print("Trade ID:", df["trade_id"].iloc[0])
print("Trade Version:", df["trade_version"].iloc[0])
print("Notional:", df["notional_amount"].iloc[0])
print("Validation Status:", df["validation_status"].iloc[0])
print("Validation Run ID:", df["validation_run_id"].iloc[0])
print("Duplicate trade/version:", df.duplicated(
    subset=["trade_id", "trade_version"]
).sum())
```
