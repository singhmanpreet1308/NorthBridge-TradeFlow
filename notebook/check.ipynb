```python

import pandas as pd

base = pd.read_csv(r"D:\STUDY\Data_Science_Courses\PROJECTS\14.NorthBridge TradeFlow\NorthBridge-TradeFlow\data\Phase_1\sample_trades.csv")
errors = pd.read_csv(r"D:\STUDY\Data_Science_Courses\PROJECTS\14.NorthBridge TradeFlow\NorthBridge-TradeFlow\data\Phase_1\seeded_errors.csv")
manifest = pd.read_csv(r"D:\STUDY\Data_Science_Courses\PROJECTS\14.NorthBridge TradeFlow\NorthBridge-TradeFlow\data\Phase_1\seeded_error_manifest.csv")
confirmations = pd.read_csv(
    r"D:\STUDY\Data_Science_Courses\PROJECTS\14.NorthBridge TradeFlow\NorthBridge-TradeFlow\data\Phase_1\independent_confirmations.csv"
)

captured = base.loc[
    base["trade_id"] == "T-1001",
    "notional_amount"
].iloc[0]

confirmed = confirmations.loc[
    confirmations["trade_id"] == "T-1001",
    "notional_amount"
].iloc[0]

assert len(base) == 50
assert len(errors) == 10
assert len(manifest) == 10
assert len(confirmations) == 50

assert captured == 100_000_000
assert confirmed == 10_000_000
assert captured != confirmed

assert (
    manifest.loc[
        manifest["fixture_trade_id"] == "T-1001",
        "expected_rule"
    ].iloc[0] == "DEMO_NOTIONAL_MISMATCH"
)

assert errors["trade_id"].isna().sum() == 1

print("Fixture integrity checks passed.")
print("T-1001 captured notional:", captured)
print("T-1001 confirmed notional:", confirmed)
print("Expected rule: DEMO_NOTIONAL_MISMATCH")
```
